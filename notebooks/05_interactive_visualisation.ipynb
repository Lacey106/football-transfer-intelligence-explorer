```python
top_edges_all = (
    club_edges
    .sort_values("transfer_count", ascending=False)
    .head(300)
    .copy()
)

G_temp = nx.Graph()

for _, row in top_edges_all.iterrows():
    G_temp.add_edge(
        row["source_club"],
        row["target_club"],
        weight=row["transfer_count"],
        total_transfer_fee=row["total_transfer_fee"]
    )

print("All top network nodes:", G_temp.number_of_nodes())
print("All top network edges:", G_temp.number_of_edges())
```

```python
largest_component_nodes = max(nx.connected_components(G_temp), key=len)

G_top = G_temp.subgraph(largest_component_nodes).copy()

print("Largest component nodes:", G_top.number_of_nodes())
print("Largest component edges:", G_top.number_of_edges())
```

```python
pos = nx.spring_layout(
    G_top,
    k=0.9,
    iterations=300,
    seed=42,
    weight="weight"
)
```

```python
edge_traces = []

max_weight = max([data["weight"] for _, _, data in G_top.edges(data=True)])

for source, target, data in G_top.edges(data=True):
    x0, y0 = pos[source]
    x1, y1 = pos[target]
    
    width = 0.5 + (data["weight"] / max_weight) * 3
    
    edge_trace = go.Scatter(
        x=[x0, x1, None],
        y=[y0, y1, None],
        line=dict(
            width=width,
            color="rgba(100,100,100,0.35)"
        ),
        hoverinfo="text",
        text=f"{source} ↔ {target}<br>Transfers: {data['weight']}<br>Total fee: {data['total_transfer_fee']}",
        mode="lines",
        showlegend=False
    )
    
    edge_traces.append(edge_trace)
```

```python
node_x = []
node_y = []
node_text = []
node_size = []
node_color = []
node_labels = []

top_label_set = set(
    club_embeddings[
        club_embeddings["club"].isin(G_top.nodes())
    ]
    .sort_values("pagerank", ascending=False)
    .head(20)["club"]
)

max_pagerank = club_embeddings["pagerank"].max()

for node in G_top.nodes():
    x, y = pos[node]
    node_x.append(x)
    node_y.append(y)
    
    row = club_embeddings[club_embeddings["club"] == node]
    
    if len(row) > 0:
        row = row.iloc[0]
        
        text = (
            f"<b>{row['club']}</b><br>"
            f"Country: {row['country']}<br>"
            f"PageRank: {row['pagerank']:.5f}<br>"
            f"Degree: {row['degree']}<br>"
            f"Weighted degree: {row['weighted_degree']}<br>"
            f"Transfers in: {row['transfers_in']}<br>"
            f"Transfers out: {row['transfers_out']}<br>"
            f"Louvain cluster: {row['louvain_cluster']}"
        )
        
        size = 7 + (row["pagerank"] / max_pagerank) * 20
        color = int(row["louvain_cluster"])
        label = row["club"] if row["club"] in top_label_set else ""
        
    else:
        text = f"<b>{node}</b>"
        size = 7
        color = 0
        label = ""
    
    node_text.append(text)
    node_size.append(size)
    node_color.append(color)
    node_labels.append(label)
```

```python
node_trace = go.Scatter(
    x=node_x,
    y=node_y,
    mode="markers+text",
    hoverinfo="text",
    text=node_labels,
    hovertext=node_text,
    textposition="top center",
    textfont=dict(size=10, color="black"),
    marker=dict(
        color=node_color,
        colorscale="Viridis",
        size=node_size,
        opacity=0.9,
        line=dict(width=1.2, color="white"),
        colorbar=dict(title="Louvain Cluster")
    ),
    showlegend=False
)
```

```python
fig_network = go.Figure(
    data=edge_traces + [node_trace],
    layout=go.Layout(
        title=dict(
            text="Largest Connected Component of Top Football Transfer Relationships",
            font=dict(size=22)
        ),
        showlegend=False,
        hovermode="closest",
        width=1100,
        height=800,
        margin=dict(b=20, l=20, r=20, t=70),
        plot_bgcolor="white",
        paper_bgcolor="white",
        xaxis=dict(
            showgrid=False,
            zeroline=False,
            showticklabels=False
        ),
        yaxis=dict(
            showgrid=False,
            zeroline=False,
            showticklabels=False
        )
    )
)

fig_network.show()
```

```python
fig_network.write_html("interactive_largest_component_transfer_network.html")

print("Saved: interactive_largest_component_transfer_network.html")
```
